```python
import pandas as pd
cols = ['상가업소번호', '상호명', '지점명', '상권업종소분류명', '시군구명', '법정동명']
df = pd.read_csv('seoul_commercial_Q3_2026.csv', usecols=cols)
df.shape
```

```python
df['상호/지점'] = df['상호명'].fillna('')+''+df['지점명'].fillna(' ').str.strip()
df['상호/지점'].head(10)
```

```python
df.drop(columns=['상호명', '지점명'], inplace=True)
df.head(20)
```

```python
#9월 기준 서울 전체 상가 수 
seoul_total = df.drop_duplicates(subset='상가업소번호')
seoul_total.shape
```

```python
#서울 내 총 카페 수
seoul_cafe_total = len(seoul_total[seoul_total['상권업종소분류명'] == '카페'])
seoul_cafe_total
```

```python
seoul_total.head()
```

```python
#서울 내 카페 중 프랜차이즈인 카페의 수
cafe_franchise = seoul_total[(seoul_total['상권업종소분류명']=='카페') & (seoul_total['상호/지점'].str.contains('점', na=False))]
cafe_franchise.tail(30)
len(cafe_franchise)
```

```python
# 서울 내 카페 중 프랜차이즈로 추정되는 카페의 비중
franchise_ratio = round((len(cafe_franchise) / seoul_cafe_total)*100,4)
franchise_ratio
```

```python
cafe_franchise['상호/지점'].value_counts().head()
```

```python
#9월 기준 서울 상권업종소분류명별 상가 수 
seoul_per_cat = df.groupby('상권업종소분류명')['상가업소번호'].count().sort_values(ascending=False)
seoul_per_cat
```

```python
# 업종별/시군구명/법정동명 별 상가 수를 'counts'라는 변수에 저장
counts = seoul_total.groupby(['상권업종소분류명', '시군구명'])['상가업소번호'].count().reset_index(name='상가수')
counts['상가수'].sum()
counts.sort_values(by=['상가수'], ascending=False)
```

```python
# '구'별 상가 수 
counts.groupby(['시군구명'])['상가수'].sum()
counts.sort_values(by='상가수', ascending=False)
```

```python
counts['구 전체 상가수']=counts.groupby(['시군구명'])['상가수'].transform('sum')
counts.sort_values(by='상가수', ascending=False)
```

```python
counts['구내 업종 비중'] = round((counts['상가수'] / counts['구 전체 상가수'])*100, 5)
counts.sort_values(by='상가수', ascending=False).head(20)
```

```python
# seoul_cafe 라는 변수를 만들어 시군구명별 카페 상가수를 반환하는 표를 넣는다. 
seoul_cafe = counts[counts['상권업종소분류명'] == '카페']
seoul_cafe.sort_values(by='구내 업종 비중', ascending=False)
```

```python
# 서울 전체 카페 비중(기준값)
seoul_cafe_share = (seoul_cafe_total / len(seoul_total)) * 100
seoul_cafe_share
print(round(seoul_cafe_share, 3))
```

```python
seoul_cafe.head()
len(seoul_cafe)
```

```python
cafe2 = seoul_cafe[['시군구명', '상가수']].rename(columns={'상가수': '구내 카페수'})
```

```python
merged = counts.merge(cafe2, on='시군구명', how='left')
merged.sort_values(by='구내 업종 비중', ascending=False)
```

```python
merged['구내 카페 비중'] = round((merged['구내 카페수' ] / merged['구 전체 상가수']) * 100, 5)
merged.sort_values(by='구내 카페수', ascending=False)
```

```python
final = merged[['시군구명', '구 전체 상가수', '구내 카페수', '구내 카페 비중']].drop_duplicates()
final
```

```python
final['서울 카페 비중'] = seoul_cafe_share
final.sort_values(by='구내 카페 비중', ascending=True)
```
